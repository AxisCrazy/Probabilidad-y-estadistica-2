```python
import numpy as np
import matplotlib.pyplot as plt

def mean(array):
    summ = 0.0
    Narray = np.size(array)
    for i in range(Narray):
        summ += array[i]
    return summ / Narray

def var(array, media):
    summ = 0.0
    Narray = np.size(array)
    for i in range(Narray):
        summ += (array[i] - media)**2
    return summ / (Narray - 1)
```

```python
Nmuestra = 100
mu=0.0 ;sigma = 1.0
data_syn = np.random.normal(mu, sigma, Nmuestra)
print(data_syn)
```

```python
Nexps = 5000
media_pob_array = np.zeros(Nexps, dtype=np.float64)

for i in range(Nexps):
    data_syn = np.random.normal(mu, sigma, Nmuestra)
    media_pob = mean(data_syn)
    varianza_pob = var(data_syn, media_pob) 
    
    media_pob_array[i] = (media_pob - mu) * np.sqrt(Nmuestra) / sigma
```

```python
fig = plt.figure(figsize=(8, 7))
z_arr = np.linspace(-3.5, 3.5, 500)
normal = 1.0 / np.sqrt(2.0 * np.pi) * np.exp(-z_arr**2 * 0.5)

plt.plot(z_arr, normal, linestyle='--', color='black')

plt.hist(media_pob_array, bins=40, density=True, histtype='step', color='blue')

plt.xlabel(r'$\overline{X}$', fontsize=14)
plt.ylabel(r'$p(\overline{X})$', fontsize=14)
plt.tick_params(axis='both', labelsize=14)
plt.show()
```
