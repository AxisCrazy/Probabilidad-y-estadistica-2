# Prueba de Normalidad (Shapiro-Wilk)

Este notebook carga los datasets `df_limpio.csv` y `conapesca_historico_2015_2025_LIMPIO.csv` y realiza la prueba de Shapiro-Wilk sobre las columnas de peso de los peces para evaluar si siguen una distribución normal.

*   **`df_limpio.csv`**: Columna `weight`
*   **`conapesca_historico_2015_2025_LIMPIO.csv`**: Columna `PESO_VIVO_KG`

```python
import pandas as pd
from scipy import stats
```

```python
# 1. Cargar los datasets
df_limpio = pd.read_csv('df_limpio.csv')
df_conapesca = pd.read_csv('conapesca_historico_2015_2025_LIMPIO.csv')

print(f'Dimensiones de df_limpio: {df_limpio.shape}')
print(f'Dimensiones de df_conapesca: {df_conapesca.shape}')
```

```python
# 2. Función para aplicar el test de Shapiro-Wilk
def aplicar_shapiro(df, columna, nombre_dataset):
    print(f'--- Resultados para {nombre_dataset} ({columna}) ---')
    
    # Extraer columna y quitar valores nulos
    datos = df[columna].dropna()
    
    # La prueba de Shapiro-Wilk no es precisa para N > 5000 (scipy muestra un Warning).
    # Si supera este límite, tomamos una muestra aleatoria de 5000.
    if len(datos) > 5000:
        print(f'El dataset es muy grande (N={len(datos)}). Usando muestra aleatoria de N=5000.')
        datos = datos.sample(n=5000, random_state=42)
        
    # Aplicar la prueba de normalidad
    stat, p = stats.shapiro(datos)
    print(f'Estadístico de prueba (W): {stat:.4f}')
    print(f'Valor-p: {p:.4e}')
    
    # Interpretación estadística
    alpha = 0.05
    if p > alpha:
        print('Conclusión: Los datos parecen provenir de una distribución normal (No se rechaza H0).\n')
    else:
        print('Conclusión: Los datos NO provienen de una distribución normal (Se rechaza H0).\n')

# 3. Ejecutar las pruebas
aplicar_shapiro(df_limpio, 'weight', 'df_limpio.csv')
aplicar_shapiro(df_conapesca, 'PESO_VIVO_KG', 'conapesca_historico_2015_2025_LIMPIO.csv')
```
