# Prueba de Normalidad (Shapiro-Wilk) - Datasets Completos

Este notebook carga los datasets `df_limpio.csv` y `conapesca_historico_2015_2025_LIMPIO.csv` y realiza la prueba de Shapiro-Wilk utilizando **todos los registros** disponibles para el peso de los peces, sin aplicar muestreo.

*   **`df_limpio.csv`**: Columna `weight`
*   **`conapesca_historico_2015_2025_LIMPIO.csv`**: Columna `PESO_VIVO_KG`

*Nota metodológica: La función `scipy.stats.shapiro` puede emitir una advertencia (UserWarning) cuando N > 5000 indicando que el valor-p puede no ser exacto, pero el estadístico W sigue calculándose para todo el volumen de datos.*

```python
import pandas as pd
import warnings
from scipy import stats

# Ignorar la advertencia de scipy para N > 5000 para mantener limpio el output
warnings.filterwarnings('ignore', category=UserWarning)
```

```python
# 1. Cargar los datasets
df_limpio = pd.read_csv('df_limpio.csv')
df_conapesca = pd.read_csv('conapesca_historico_2015_2025_LIMPIO.csv')

print(f'Dimensiones de df_limpio: {df_limpio.shape}')
print(f'Dimensiones de df_conapesca: {df_conapesca.shape}')
```

```python
# 2. Función para aplicar el test de Shapiro-Wilk (Dataset Completo)
def aplicar_shapiro_completo(df, columna, nombre_dataset):
    print(f'\n--- Resultados para {nombre_dataset} ({columna}) ---')
    
    # Extraer columna y quitar valores nulos
    datos = df[columna].dropna()
    print(f'Evaluando la totalidad de los datos (N={len(datos)})...')
    
    # Aplicar la prueba de normalidad a TODO el dataset
    stat, p = stats.shapiro(datos)
    print(f'Estadístico de prueba (W): {stat:.4f}')
    print(f'Valor-p: {p:.4e}')
    
    # Interpretación estadística
    alpha = 0.05
    if p > alpha:
        print('Conclusión: Los datos parecen provenir de una distribución normal (No se rechaza H0).')
    else:
        print('Conclusión: Los datos NO provienen de una distribución normal (Se rechaza H0).')

# 3. Ejecutar las pruebas con todos los datos
aplicar_shapiro_completo(df_limpio, 'weight', 'df_limpio.csv')
aplicar_shapiro_completo(df_conapesca, 'PESO_VIVO_KG', 'conapesca_historico_2015_2025_LIMPIO.csv')
```
